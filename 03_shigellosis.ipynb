### Daten hereinladen

```python
# Daten laden --> read_csv / read_excel
# Falls mehrere sheets --> sheet_name="sheet"


data = pd.read_csv("data/shigellosis/data.csv")

data
```

### Variablen

```python
data.columns
```

* valueCategory: Was gemessen wird --> Hier: Anzahl gemeldete Fälle
* temporal: Welcher Zeitabschnitt --> Jahr und Monat
* temporal_type: Typ der Zeit --> Monat
* georegion: Kanton oder Land --> hier Anfang Land später Kanton
* georegion_type: Typ der Georegion --> Land/Kantonal
* agegroup: Alterskategorie
* agegroup_type: Zuordnung zur Oberthema --> bei allen gleich
* sex: Geschlecht
* value: Wert basierend auf vorhergehende Kategorien
* pop: Bevölkerung zum Zeitpunkt der Erhebung
* incValue: neuaufgetrettene Fälle innerhalb der Gruppe und Zeit
* valueYearCumSum: Gesamter Wert über das Jahr
* valueMean3y: Durchschnittswert der letzten drei Jahre
* trend: ist der Trend abnehmend(falling), gleich(stagnant) oder steigend(rising)
* prct: prozentualer Anteil der Fälle in den Kategorien und Zeiteinheit
* deltaValue: Unterschied des value zum vorherigen Wert
* deltaIncValue: Unterschied des incValue zum vorherigen Wert
* incValuePrctChange: Prozentualer Unterschied zum vorherigen Wert
* valueMean5y: Durchschnittlicher Wert der letzten 5 Jahre
* dataComplete: ist dieser Datenpunkt komplett

### Datensatz für Cases

--> Die Durchschnittswerte sind als Einzelne Datenpunkte erhoben

```python
data_def = data[data["valueCategory"] == "cases"]

data_def
```

### Datensatz für Agegroup

```python
data_med = data[data["valueCategory"] == "cases_10y_mean"]

data_med
```

### Datensatz für Geschlecht

```python
data_sex = data[data["valueCategory"] == "cases_10y_cumsum"]

data_sex
```

### Visuelle Analyse

```python
#Gruppieren nach georegion und Werte summieren
counts = (
    data_def.groupby("georegion")
        .agg(sum_value=("value", "sum"))
        .reset_index()
)

#Bar Chart
bar_georegion = alt.Chart(counts).mark_bar(color="aqua").encode(
    x=alt.X("georegion", title="Georegion"),
    y=alt.Y("sum_value", title="Total Cases")
).properties(
    title="Shigellosis Georegion"
)

# Textlabels
text = bar_georegion.mark_text(
    align="center",
    dy=-5,
    size=10
).encode(
    text="sum_value:Q"
)

# Kombinieren
bar_georegion = bar_georegion + text
bar_georegion
```

* Teilweise CH, teilweise CHFL

```python
# Ohne Kantone --> ~
data_cantons = data_def[~data_def["georegion"].isin(["CH", "CHFL"])]

#Gruppieren nach georegion und Werte summieren
counts = (
    data_cantons.groupby("georegion")
        .agg(sum_value=("value", "sum"))
        .reset_index()
)

#Bar Chart
bar_georegion = alt.Chart(counts).mark_bar(color="aqua").encode(
    x=alt.X("georegion", title="Georegion"),
    y=alt.Y("sum_value", title="Total Cases")
).properties(
    title="Shigellosis Georegion 2013-2025"
)

# Textlabels
text = bar_georegion.mark_text(
    align="center",
    dy=-5,
    size=10
).encode(
    text="sum_value:Q"
)

# Kombinieren
bar_georegion = bar_georegion + text
bar_georegion
```

```python
#Gruppieren nach georegion und Werte summieren
counts = (
    data_def.groupby("temporal")
        .agg(sum_value=("value", "sum"))
        .reset_index()
)

#Bar Chart
bar_temporal = alt.Chart(counts).mark_bar(color="aqua").encode(
    x=alt.X("temporal", title="Temporal"),
    y=alt.Y("sum_value", title="Cases")
).properties(
    title="Shigellosis Temporal"
)

# Textlabels
text = bar_temporal.mark_text(
    align="center",
    dy=-5,
    size=10
).encode(
    text="sum_value:Q"
)

# Kombinieren
bar_temporal = bar_temporal + text
bar_temporal
```

* Datensatz aufteilen in monatlich und Jährlich

```python
data_1 = data_def.copy()

data_year = data_1[data_1["temporal_type"] == "year"]

data_year
```

```python
# Ohne Kantone
data_year_CH = data_year[data_year["georegion"].isin(["CH", "CHFL"])]

bar_year = alt.Chart(data_year_CH).mark_bar().encode(
    x=alt.X("temporal", title="Temporal"),
    y=alt.Y("value", title="Cases"),
    color=alt.Color("incValue")
).properties(
    title="Shigellosis Year"
)

bar_year 
```

* Peak in 2018
* Meklich weniger Fälle in 2020
* Erneuter Anstieg am 2022

```python
data_2 = data_def.copy()

data_month = data_2[data_2["temporal_type"] == "month"]

data_month
```

```python
bar_month = alt.Chart(data_month).mark_bar().encode(
    x=alt.X("temporal:N", title="Temporal"),
    y=alt.Y("value:Q", title="Cases"),
    color=alt.Color("trend:N")
).properties(
    title="Shigellosis Month"
)

bar_month
```

* sehr unterschiedlich im Jahr
* merklich weniger Fälle 2020

```python
bar_age = alt.Chart(data_med).mark_bar().encode(
    x=alt.X("agegroup", title="Agegroup"),
    y=alt.Y("incValue", title="Cases"),
    color=alt.Color("temporal")
).properties(
    title="Salmonellosis Altersgruppen 2022 - 2025"
)

bar_age
```

```python
data_age_2025 = data_med[data_med["temporal"] == "2025"]

bar_age_2 = alt.Chart(data_age_2025).mark_bar().encode(
    x=alt.X("agegroup", title="Agegroup"),
    y=alt.Y("incValue", title="Cases")
).properties(
    title="Salmonellosis Altersgruppen 2025"
)

bar_age_2
```
