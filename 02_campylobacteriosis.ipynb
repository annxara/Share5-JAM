### Variablen

```python
data.columns
```

* valueCategory: Was gemessen wird --> Hier: Anzahl gemeldete Fälle
* temporal: Welcher Zeitabschnitt --> Jahr und Monat
* temporal_type: Typ der Zeit --> Monat
* georegion: Kanton oder Land --> hier Anfang Land später Kanton
* georegion_type: Typ der Georegion --> Land/Kantonal
* agegroup: Alterskategorie
* agegroup_type: Zuordnung zur Oberthema --> bei allen gleich
* sex: Geschlecht
* value: Wert basierend auf vorhergehende Kategorien
* pop: Bevölkerung zum Zeitpunkt der Erhebung
* incValue: neuaufgetrettene Fälle innerhalb der Gruppe und Zeit
* valueYearCumSum: Gesamter Wert über das Jahr
* valueMean3y: Durchschnittswert der letzten drei Jahre
* trend: ist der Trend abnehmend(falling), gleich(stagnant) oder steigend(rising)
* prct: prozentualer Anteil der Fälle in den Kategorien und Zeiteinheit
* deltaValue: Unterschied des value zum vorherigen Wert
* deltaIncValue: Unterschied des incValue zum vorherigen Wert
* incValuePrctChange: Prozentualer Unterschied zum vorherigen Wert
* valueMean5y: Durchschnittlicher Wert der letzten 5 Jahre
* dataComplete: ist dieser Datenpunkt komplett

### Visuelle Analyse

```python
# Ohne Kantone --> ~
data_cantons = data_def[~data_def["georegion"].isin(["CH", "CHFL"])]

#Gruppieren nach georegion und Werte summieren
counts = (
    data_cantons.groupby("georegion")
        .agg(sum_value=("value", "sum"))
        .reset_index()
)

#Bar Chart
bar_georegion = alt.Chart(counts).mark_bar(color="aqua").encode(
    x=alt.X("georegion", title="Georegion"),
    y=alt.Y("sum_value", title="Total Cases")
).properties(
    title="Campylobacteriosis Georegion 2013-2025"
)

# Textlabels
text = bar_georegion.mark_text(
    align="center",
    dy=-5,
    size=10
).encode(
    text="sum_value:Q"
)

# Kombinieren
bar_georegion = bar_georegion + text
bar_georegion
```

* ZH am höchsten --> grösster Kanton
* kleine Kantone --> weniger Fälle

```python
# nur 2025
data_cantons_2025 = data_cantons[data_cantons["temporal"] == "2025"]

#Gruppieren nach georegion und Werte summieren
counts = (
    data_cantons_2025.groupby("georegion")
        .agg(sum_value=("value", "sum"))
        .reset_index()
)

#Bar Chart
bar_georegion_2025 = alt.Chart(counts).mark_bar(color="aqua").encode(
    x=alt.X("georegion", title="Georegion"),
    y=alt.Y("sum_value", title="Total Cases")
).properties(
    title="Campylobacteriosis Georegion 2025"
)

# Textlabels
text = bar_georegion_2025.mark_text(
    align="center",
    dy=-5,
    size=10
).encode(
    text="sum_value:Q"
)

# Kombinieren
bar_georegion_2025 = bar_georegion_2025 + text
bar_georegion_2025
```

### Durchschnittsberechnungen & -visualisierungen

```python
# Ohne Kantone --> ~
data_cantons = data_def[~data_def["georegion"].isin(["CH", "CHFL"])]

#Gruppieren nach georegion und Werte summieren
counts_mean = (
    data_cantons.groupby("georegion")
        .agg(mean_value=("value", "mean")).round(0)
        .reset_index()
)

#Bar Chart
bar_georegion = alt.Chart(counts_mean).mark_bar(color="aqua").encode(
    x=alt.X("georegion", title="Georegion"),
    y=alt.Y("mean_value", title="Mean Cases")
).properties(
    title="Campylobacteriosis Georegion 2013-2025 Mean"
)

# Textlabels
text = bar_georegion.mark_text(
    align="center",
    dy=-5,
    size=10
).encode(
    text="mean_value:Q"
)

# Kombinieren
bar_georegion = bar_georegion + text
bar_georegion
```

```python
# Jahr und Monat trennen
data_month[["year", "month"]] = data_def["temporal"].str.split("-", expand=True)

data_month
```

```python
counts_mean = (
    data_month.groupby("month")
        .agg(mean_value=("value", "mean")).round(0)
        .reset_index()
)

bar_month = alt.Chart(counts_mean).mark_bar().encode(
    x=alt.X("month:N", title="Month"),
    y=alt.Y("mean_value:Q", title="Cases")
).properties(
    title="Campylobacteriosis Month 2013-2026 Mean"
)

# Textlabels
text = bar_month.mark_text(
    align="center",
    dy=-5,
    size=10
).encode(
    text="mean_value:Q"
)

bar_month = bar_month + text
bar_month
```

```python
counts_mean = (
    data_med.groupby("agegroup")
        .agg(mean_value=("incValue", "mean")).round(0)
        .reset_index()
)

bar_age = alt.Chart(counts_mean).mark_bar().encode(
    x=alt.X("agegroup", title="Agegroup"),
    y=alt.Y("mean_value", title="Cases")
).properties(
    title="Campylobacteriosis Altersgruppen 2022 - 2025 Mean IncValue"
)

# Textlabels
text = bar_age.mark_text(
    align="center",
    dy=-5,
    size=10
).encode(
    text="mean_value:Q"
)

bar_age = bar_age + text
bar_age
```

### Auf 100'000 Personen gerechnet

```python
data_cantons
```

```python
data_cantons_100 = data_cantons.copy()

# Wert pro 100000 einfügen
data_cantons_100["value_100000"] = data_cantons_100["value"] / data_cantons_100["pop"] * 100000

data_cantons_100["value_100000"] = data_cantons_100["value_100000"].round(1)

data_cantons_100
```

```python
# nur 2025
data_cantons_100_2025 = data_cantons_100[data_cantons_100["temporal"] == "2025"]

#Bar Chart
bar_georegion = alt.Chart(data_cantons_100_2025).mark_bar(color="aqua").encode(
    x=alt.X("georegion", title="Georegion"),
    y=alt.Y("value_100000", title="Mean Cases")
).properties(
    title="Campylobacteriosis Georegion 2025 pro 100000"
)

# Textlabels
text = bar_georegion.mark_text(
    align="center",
    dy=-5,
    size=10
).encode(
    text="value_100000:Q"
)

# Kombinieren
bar_georegion = bar_georegion + text
bar_georegion
```
